### Diagramme a barres :

```python
#Créer une visualisation des données du graphique à barres qui montre les dépenses des clients par pays,

dépenses_client_par_pays=df_sell4all.groupby("Country")["Customer spendings"].median()

pays_depense=dépenses_client_par_pays.index
custmer_depense=dépenses_client_par_pays.values
plt.figure(figsize=(12, 6))
plt.bar(pays_depense , custmer_depense  , color='#2980B9')

# pour mettre un titre au notre diagrammes à barres :
plt.title("Dépenses des clients par pays")
plt.xlabel("Pays")
plt.xticks(rotation=95)
plt.ylabel("Dépenses")
plt.show()
```
